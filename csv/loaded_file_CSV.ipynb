**Pima Indan DIabeti Database**

*this is a database for diabetic patients*

*This database looks at whether a patient has diabetes, based on 8 body measurements. It covers 768 adult women of Pima Indian heritage.*

Pregnancies - Number of times pregnant.

Glucose - Blood sugar level 2 hours into an oral glucose tolerance test.

BloodPressure - Diastolic blood pressure in mm Hg.

SkinThickness - Triceps skin fold thickness in mm.

Insulin - Blood insulin level 2 hours after the glucose test.

BMI - Body mass index, weight in kg divided by height in meters squared.

DiabetesPedigreeFunction - Score of family history of diabetes.

Age - Age in years.

Outcome - Diabetes status, 1 means diabetes and 0 means no diabetes.

**Breast Cancer Wisconsin Database**

This database looks at whether a breast tumor is benign or malignant, based on 9 cell measurements. It covers 699 samples.

Id - Sample code number.

Cl.thickness - Clump thickness, scored 1 to 10.

Cell.size - Uniformity of cell size, scored 1 to 10.

Cell.shape - Uniformity of cell shape, scored 1 to 10.

Marg.adhesion - Marginal adhesion, scored 1 to 10.

Epith.c.size - Single epithelial cell size, scored 1 to 10.

Bare.nuclei - Bare nuclei, scored 1 to 10.

Bl.cromatin - Bland chromatin, scored 1 to 10.

Normal.nucleoli - Normal nucleoli, scored 1 to 10.

Mitoses - Mitoses, scored 1 to 10.

Class - Diagnosis, 0 means benign and 1 means malignant.
