**Real Estate Valuation**

**Coloumns**

* `No`: row index
* `X1 transaction date`: transaction date as decimal year, e.g. 2012.917
* `X2 house age`: house age in years
* `X3 distance to the nearest MRT station`: distance in meters to nearest MRT station
* `X4 number of convenience stores`: count of nearby convenience stores

* `X6 longitude`: longitude in degrees

**Default of Credit Card Clients**

**About:** Payment records for 30,000 credit card clients in Taiwan. Covers credit limit, demographics, six months of payment history, bill amounts, and payment amounts. Target: whether the client defaulted the next month.

**Columns**

* `ID`: row index
* `LIMIT_BAL`: credit limit, New Taiwan Dollar
* `SEX`: 1 = male, 2 = female
* `EDUCATION`: 1 = graduate school, 2 = university, 3 = high school, 4 = other, 5/6/0 = unknown
* `MARRIAGE`: 1 = married, 2 = single, 3 = other, 0 = unknown
* `AGE`: age in years
* `PAY_0`: repayment status, September
* `PAY_2`: repayment status, August
* `PAY_3`: repayment status, July
* `PAY_4`: repayment status, June
* `PAY_5`: repayment status, May
* `PAY_6`: repayment status, April
* `BILL_AMT1` to `BILL_AMT6`: bill statement amount, per month, New Taiwan Dollar
* `PAY_AMT1` to `PAY_AMT6`: previous payment amount, per month, New Taiwan Dollar
* `default payment next month`: 1 = default, 0 = no default

```python
import pandas as pd

db2_credit=pd.read_excel('https://archive.ics.uci.edu/ml/machine-learning-databases/00350/default%20of%20credit%20card%20clients.xls')
db2_credit.head(11)
```
