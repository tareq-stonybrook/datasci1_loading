***JSON FILES***

**Cars**

*Specs for 406 cars. Model years 1970 to 1982. Origins: USA, Europe, Japan. Covers fuel economy and engine traits.*

**Columns**

* `Name`: car make and model
* `Miles_per_Gallon`: fuel economy
* `Cylinders`: cylinder count
* `Displacement`: engine size
* `Horsepower`: engine power
* `Weight_in_lbs`: vehicle weight in pounds
* `Acceleration`: acceleration measure
* `Year`: model year, stored as date
* `Origin`: USA, Europe, or Japan

```python
import pandas as pd

db1_cars=pd.read_json('https://raw.githubusercontent.com/vega/vega-datasets/main/data/cars.json')
db1_cars.head(11)
```

**Penguins**

**About:** Measurements for 344 penguins. Three species: Adelie, Chinstrap, Gentoo. Collected in the Palmer Archipelago, Antarctica. Covers beak size, flipper size, body mass, and sex.

**Columns**

* `Species`: Adelie, Chinstrap, or Gentoo

* `Island`: Torgersen, Biscoe, or Dream

* `Beak Length (mm)`: beak length in millimeters

* `Beak Depth (mm)`: beak depth in millimeters

* `Flipper Length (mm)`: flipper length in millimeters

* `Body Mass (g)`: body mass in grams

* `Sex`: MALE or FEMALE

```python
import pandas as pd

db2_peng=pd.read_json('https://raw.githubusercontent.com/vega/vega-datasets/main/data/penguins.json')
db2_peng.head(11)
```

**Gapminder**

**About:** Country level demographic data for 62 countries. Years 1955 to 2005 in 5 year steps. Covers population, life expectancy, and fertility. Source: Gapminder Foundation.


**Columns**

* `year`: year of record, 1955 to 2005 in 5 year steps
* `country`: country name
* `cluster`: region group, coded 0 to 5
* `pop`: population of country
* `life_expect`: life expectancy in years
* `fertility`: average children per woman



```python
import pandas as pd

db3_gapminder=pd.read_json('https://raw.githubusercontent.com/vega/vega-datasets/main/data/gapminder.json')
db3_gapminder.head(11)
```
